```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')

from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-20b", groq_api_key=groq_api_key)
```

```python
llm
```

```python
from langchain.tools import tool

@tool
def get_weather(area:str) -> str:
    """Get the weather of the area"""
    return f"The weather of the {area} is Sunny"

model_with_tools = llm.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("Whats the weather in Korutla?")
```

```python
print(response)
```

### Tool Execution Steps

```python
#step 1: model generates the calls
messages = [{"role":"user","content":"Whats the weather in Boston"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)
```

```python
ai_msg
```

```python
messages
```

```python
# Step:2 Execute the tools and collect results
for tool_call in ai_msg.tool_calls:
    #execute the tool with the generated arguments
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)
```

```python
messages
```

```python
#step 3: Pass results back to model for final response
final_response = model_with_tools.invoke(messages)
print(final_response.text)
```
