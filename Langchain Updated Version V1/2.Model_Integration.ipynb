## Streaming and Batch

### Streaming

Most models can stream thier output while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling Stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time:

```python
llm.invoke("Say me about Artificial Intelligence")
```

```python
llm.stream("Artificial Intellingence")
```

```python
for chunk in llm.stream("Say me about Artificial Intelligence"):
    print(chunk.text, end="|", flush = True)
```

```python
for chunk in llm.stream("How AI will rule the world in future"):
    print(chunk.text,  end="|", flush = True)
```

```python
llm.invoke("How AI will rule the world in future")
```

### Batch

Batching is a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = llm.batch([
    "Why do you like mumbai Indians",
    "How AI will impact in future",
    "What are key steps of success"
], config={"max_concurrency":5})
```
