```python
#PDF Loader
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("Sumanth_bandi_Resume.pdf")

pdf_doc = loader.load()
```

```python
pdf_doc
```

```python
loader2 = PyPDFLoader("Sumanth_bandi_Resume.pdf").load()
```

```python
loader2
```

# SPlitting of the text

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 30)

docs = splitter.split_documents(loader2)
```
