```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.document_loaders import PyPDFLoader
loader  = PyPDFLoader("Sumanth_Bandi_Resume.pdf")
documents = loader.load()
print(documents)
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
split_docs = splitter.split_documents(documents)
```

```python
split_docs
```

### Embeddings

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model = "text-embedding-3-large")
```

```python
query = "What is the experience of Sumanth Bandi in Python?"

query_embedding = embeddings.embed_query(query)
query_embedding
```

```python
len(query_embedding)
```

### Vector Storage and embedding together

```python
from langchain.vectorstores import Chroma
vector_db = Chroma.from_documents(split_docs, embeddings)
```

```python
vector_db
```

```python
#retrieving the data from the vector database
query = "What is the experience of Sumanth Bandi in Python?"

retreived_result = vector_db.similarity_search(query)
```

```python
print(retreived_result)
```
