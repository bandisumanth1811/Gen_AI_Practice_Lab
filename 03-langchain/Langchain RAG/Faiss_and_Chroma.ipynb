```python
from langchain_community.document_loaders import TextLoader
Loader = TextLoader("speech.txt", encoding="utf-8")
documents = Loader.load()
```

```python
documents
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 100)
split_docs = splitter.split_documents(documents)
```

```python
split_docs
```

## Embedding 

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model= "text-embedding-3-small")
```

```python
from langchain.vectorstores import FAISS    
```

```python
db = FAISS.from_documents(split_docs, embeddings)
```

```python
db
```

```python
query = "What did APJ Abdul Kalam say?"
```

```python
retrived_query = db.similarity_search(query)
```

```python
retrived_query[0]
```

```python
retrived_query[0].page_content
```

### As A Retriever

```python
retriever = db.as_retriever()
```

```python
doc_query = retriever.invoke(query)
doc_query
```

### Similarity Search with Score

```python
query_score = db.similarity_search_with_score(query)
query_score
```

```python
# as a vector

embedding_vector = embeddings.embed_query(query)
embedding_vector
```

```python
query_vector = db.similarity_search_by_vector(embedding_vector)
query_vector
```

```python
## Saving the vector database
db.save_local("faiss_index")
```

```python
## Loading the vector database

new_db = FAISS.load_local("faiss_index", embeddings, allow_dangerous_deserialization= True)
```

```python
new_db.similarity_search(query)
```

### Lets implement same thing using chroma db

```python
from langchain.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain.vectorstores import Chroma
```

```python
Pdf_loader = PyPDFLoader("Sumanth_Bandi_Resume.pdf")
pdf_docs = Pdf_loader.load()
```

```python
pdf_docs
```

```python
Pdf_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
pdf_split_docs = Pdf_splitter.split_documents(pdf_docs)
pdf_split_docs
```

```python
pdf_embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
Chroma_db = Chroma.from_documents(pdf_split_docs, pdf_embeddings)
Chroma_db
```

```python
pdf_query = "Where does Sumanth Bandi is working now?"
```

```python
Chroma_db.similarity_search(pdf_query)
```

```python
Chroma_db.similarity_search(query)
```

```python
#Saving in local
Chroma_db = Chroma.from_documents(pdf_split_docs, pdf_embeddings, persist_directory="./chroma_db")
```

```python
#Loading it again

chroma_db2 = Chroma(persist_directory="./chroma_db", embedding_function = pdf_embeddings)
```

```python
chroma_db2.similarity_search(pdf_query)
```

### LLM

```python
from langchain_openai import OpenAI
llm = OpenAI(model = "gpt-4o-mini")
```

```python
llm.invoke("Where will be world in AI in 2030")
```

```python
from langchain.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system","You are an expert AI Assistant, Answer the questions based on the users input"),
    ("user","{input}") 
])
prompt
```

```python
chain = prompt|llm
```

```python
response = chain.invoke({"input":"Are you giving the response from prompt template"})
```

```python
response
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
chain = prompt|llm|parser
```

```python
response2 = chain.invoke({"input":"Do Deep Learning Founder even doesnt know how its created, or was not able to capture how things work internally?"})
```
